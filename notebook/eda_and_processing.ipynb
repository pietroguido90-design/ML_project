```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import os
```

## Define file paths using raw strings for Windows paths

```python
red_wine_path = r"C:\Users\pietr\Desktop\IronHack\Machine learning project\data\raw\winequality-red.csv"
white_wine_path = r"C:\Users\pietr\Desktop\IronHack\Machine learning project\data\raw\winequality-white.csv"
output_dir = r"C:\Users\pietr\Desktop\IronHack\Machine learning project\data\processed"
output_file = os.path.join(output_dir, "wine_quality_combined.csv")
```

## Load local CSV files (UCI wine quality datasets use semicolon ';' as a separator)

```python
df_red = pd.read_csv(red_wine_path, sep=';')
df_white = pd.read_csv(white_wine_path, sep=';')
```

## Add a categorical column to identify wine type

```python
df_red['type'] = 'red'
df_white['type'] = 'white'
```

## Combine red and white wine datasets into a single DataFrame

```python
df_wine = pd.concat([df_red, df_white], ignore_index=True)
```

## Standardize column names (lower case and replace spaces with underscores)

```python
df_wine.columns = df_wine.columns.str.lower().str.replace(' ', '_')
```

## Exploratory Data Analysis & Quality Checks

```python
print("\n--- EDA & Data Quality Checks ---")
print(f"Dataset shape: {df_wine.shape}")
```

## Check missing values

```python
missing_vals = df_wine.isnull().sum().sum()
print(f"Total missing values: {missing_vals}")
```

## Check duplicate rows

```python
duplicates = df_wine.duplicated().sum()
print(f"Duplicate rows found: {duplicates}")
```

## Create binary target for classification (1 for high quality >= 7, 0 otherwise)

```python
df_wine['quality_label'] = (df_wine['quality'] >= 7).astype(int)
```

```python
print("\nTarget Class Distribution:")
print(df_wine['quality_label'].value_counts(normalize=True))
```

##  Ensure output directory exists and save processed dataset

```python
os.makedirs(output_dir, exist_ok=True)
df_wine.to_csv(output_file, index=False)
```

## Display dataset dimensions and initial rows

```python
print(f"Combined dataset successfully created and saved at:\n{output_file}\n")
print(f"Total rows and columns: {df_wine.shape}")
print(f"Distribution by wine type:\n{df_wine['type'].value_counts()}\n")
print(f"Distribution by quality label (Class 1 = High Quality >= 7):\n{df_wine['quality_label'].value_counts(normalize=True).round(3)}")

df_wine.head()
```

## Encoding categorical feature 'type' (0 = red, 1 = white)

```python
df_wine['type_code'] = (df_wine['type'] == 'white').astype(int)
```

## Separate features (X) and target (y)

```python
X = df_wine.drop(columns=['quality', 'quality_label', 'type'])
y = df_wine['quality_label']
```

## Train-Test Split with stratification

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=17, stratify=y)

print(f"Training set shape: {X_train.shape}")
print(f"Testing set shape: {X_test.shape}")
```

## Feature Scaling

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Convert back to DataFrame for ease of use

```python
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=X.columns)

print("\nProcessing complete! Data is ready for machine learning model training.")
```
